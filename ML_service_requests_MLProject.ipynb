```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
service_requests_311.columns = (
    service_requests_311.columns
    .str.strip()
    .str.replace('(', "", regex=False)
    .str.replace(" ", "_")
    .str.replace("/", "")
    .str.replace(")", "", regex=False)

)
service_requests_311.columns.tolist()
```

```python
#service_requests_311['Closed_Date'] = (
#    service_requests_311["Closed_Date"]
#    .str.replace(' AM', "", regex=False)
#    .str.replace(' PM', "", regex=False))
#service_requests_311['Created_Date'] = (
#    service_requests_311["Created_Date"]
#    .str.replace(' AM', "", regex=False)
#    .str.replace(' PM', "", regex=False))
service_requests_311["Closed_Date"] = pd.to_datetime(
    service_requests_311["Closed_Date"],
    errors="coerce"
)
service_requests_311["Created_Date"] = pd.to_datetime(
    service_requests_311["Created_Date"],
    errors="coerce"
)

service_requests_311["Closed_hour"] = service_requests_311["Closed_Date"].dt.hour
service_requests_311["Closed_day"] = service_requests_311["Closed_Date"].dt.day
service_requests_311["Closed_month"] = service_requests_311["Closed_Date"].dt.month
service_requests_311["Closed_year"] = service_requests_311["Closed_Date"].dt.year

service_requests_311["Created_hour"] = service_requests_311["Created_Date"].dt.hour
service_requests_311["Created_day"] = service_requests_311["Created_Date"].dt.day
service_requests_311["Created_month"] = service_requests_311["Created_Date"].dt.month
service_requests_311["Created_year"] = service_requests_311["Created_Date"].dt.year

service_requests_311["Closure_duration_days"]= (
    service_requests_311["Closed_Date"] - service_requests_311["Created_Date"]
).dt.days

service_requests_311[["Closed_Date", "Closed_hour", "Closed_day", "Closed_month","Closed_year","Closure_duration_days"]].head()

```

```python
service_requests_311.drop(columns=[
    "Unique_Key","day_of_week	","Closure_duration"
], inplace=True,errors="ignore")
service_requests_311.head()
```

```python
missing_counts = service_requests_311.isna().sum().sort_values(ascending=False)

missing_counts[missing_counts > 0]
```

```python
# Separate numerical and categorical columns
numerical_cols = service_requests_311.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = service_requests_311.select_dtypes(exclude=["number"]).columns.tolist()

print("Numerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)

# Display preview
display(service_requests_311[numerical_cols + categorical_cols].head())
```

```python
#These values will be reused to fill missing values in dataset, ensuring no data leakage.
# Numerical: use median (robust to outliers)
numerical_median_values = service_requests_311[numerical_cols].median()

# Categorical: use most frequent (mode)
categorical_mode_values = service_requests_311[categorical_cols].mode().iloc[0]

print("Numerical imputation values:")
print(numerical_median_values)
print("\nCategorical imputation values:")
print(categorical_mode_values)
# Numerical
service_requests_311[numerical_cols] = service_requests_311[numerical_cols].fillna(numerical_median_values)

# Categorical
service_requests_311[categorical_cols] = service_requests_311[categorical_cols].fillna(categorical_mode_values)
service_requests_311.drop(columns=[
    "Due_Date"
], inplace=True,errors="ignore")
# Verify no missing values remain
still_missing = service_requests_311.isna().sum().sort_values(ascending=False)
still_missing[still_missing > 0]
print("Missing after imputation:", still_missing)

```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import pandas as pd
import scipy.sparse as sp

# Automatically define numerical and categorical columns from the dataframe
numerical_cols = service_requests_311.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = service_requests_311.select_dtypes(exclude=["number"]).columns.tolist()

# Identify and drop high-cardinality string columns to prevent a memory crash
high_cardinality_cols = [
    'Created_Date',
    'Closed_Date',
    'Street_Name',
    'Resolution_Description',
    'Resolution_Action_Updated_Date',
    'X_Coordinate_State_Plane',
    'Y_Coordinate_State_Plane'
]

# Filter to keep only informative, low-to-mid cardinality categorical columns
categorical_cols_clean = [col for col in categorical_cols if col not in high_cardinality_cols]
print("Categorical columns being encoded:", categorical_cols_clean)

# 1. Initialize and apply OneHotEncoder with sparse_output=True to keep memory footprint minimal
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=True)
encoded_cat_sparse = encoder.fit_transform(service_requests_311[categorical_cols_clean])

# 2. Scale the numerical features (excluding Due_Date)
scaler = StandardScaler()
numerical_cols_clean = [col for col in numerical_cols if col != 'Due_Date']
scaled_num_array = scaler.fit_transform(service_requests_311[numerical_cols_clean])

# 3. Combine them horizontally as a memory-efficient sparse matrix
scaled_num_sparse = sp.csr_matrix(scaled_num_array)
processed_dataset_sparse = sp.hstack([scaled_num_sparse, encoded_cat_sparse])

print("Preprocessing successfully completed!")
print("Final Sparse Matrix Shape:", processed_dataset_sparse.shape)
print("Session saved from crashing using sparse matrix structures.")
```

```python
# Save datasets (CSV)
# =========================

# 1. Recreate column names to structure our DataFrame
# Get the feature names from the fitted encoder
encoded_cols = encoder.get_feature_names_out(categorical_cols_clean)

# Combine numerical columns and encoded categorical columns
all_feature_cols = list(numerical_cols_clean) + list(encoded_cols)

# 2. Convert the sparse matrix to a pandas DataFrame
# We convert it to a dense format (.toarray()) just for the export stage
processed_df = pd.DataFrame(
    processed_dataset_sparse.toarray(),
    columns=all_feature_cols
)

# 3. Save the DataFrame to a CSV file
processed_df.to_csv("processed_service_requests_dataset.csv", index=False)

print("Dataset successfully converted and saved as 'processed_service_requests_dataset.csv'!")
```
